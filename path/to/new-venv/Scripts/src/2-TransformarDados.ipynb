```python
import pandas as pd
import zipfile
import os
from tabula import read_pdf
import jpype

# 2.1. Extraindo os dados da tabela do PDF
pdf_path = "https://www.gov.br/ans/pt-br/acesso-a-informacao/participacao-da-sociedade/atualizacao-do-rol-de-procedimentos/Anexo_I_Rol_2021RN_465.2021_RN627L.2024.pdf"
tables = read_pdf(pdf_path, pages='all', multiple_tables=True)

# Concatenando todas as tabelas extraídas em um único DataFrame
df = pd.concat(tables, ignore_index=True)
```

```python

# 2.4. Substituindo as abreviações OD e AMB pelas descrições completas
df.replace({'OD': 'Outros Despesas', 'AMB': 'Ambulatorial'}, inplace=True)
```

```python
# 2.2. Salvando os dados em um arquivo CSV
csv_filename = "rol_procedimentos.csv"
df.to_csv(csv_filename, index=False)
```

```python
# 2.3. Compactando o CSV em um arquivo ZIP
zip_filename = "Teste_Matheus_Silva_Correia.zip"
with zipfile.ZipFile(zip_filename, 'w') as zipf:
    zipf.write(csv_filename)

# Remover o arquivo CSV após compactação, se desejado
os.remove(csv_filename)

print("Extração, salvamento e compactação concluídos!")
```
